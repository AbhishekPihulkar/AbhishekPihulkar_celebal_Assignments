# 🧠 Single Agent Pipeline Project

## Problem Statement
Build a **Single-Agent Smart Assistant** that:
- Understands user queries
- Routes tasks based on intent
- Uses tools when required
- Returns structured JSON output

### The agent should handle:
- Math queries → Calculator Tool
- Keyword extraction → Keyword Tool
- General queries → Direct response

---
### 🛠️ What You Need to Implement
- Agent logic
- Conditional routing
- Tool integration
- Basic error handling

### 🚀 Bonus
- Improve routing
- Add logging
- Add more tools


````python
#Added logging
import re, logging
logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s", datefmt="%H:%M:%S")
logger = logging.getLogger("agent")
````

````python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
````

````python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
````

````python
# 🛠️ TOOL 3: Word Counter(Bonus)
def word_counter(text: str) -> dict:
    """Return word and character counts for a piece of text."""
    try:
        return {"word_count": len(text.split()), "char_count": len(text)}
    except Exception:
        return {}
````

````python
# 🛠️ TOOL 4 : Prime Checker(Bonus)

def is_prime(n: int) -> bool:
    """Check whether a number is prime."""
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
````

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

````python
# 🤖 AGENT FUNCTION
# Improved Routing
_MATH_PATTERN = re.compile(r"\d+\s*[\+\-*/%]\s*\d+")

def agent(query: str):
    query_lower = query.lower()

    try:
        if "calculate" in query_lower or _MATH_PATTERN.search(query_lower):
            expression = query_lower.replace("calculate", "").strip()
            logger.info(f"query='{query}' -> intent='calculation'")
            result = calculator(expression)
            if result == "Error in calculation":
                return {"type": "error", "result": result}
            return {"type": "calculation", "result": result}

        elif "keywords" in query_lower:
            text = query_lower.replace("extract keywords from", "").replace("keywords", "").strip()
            logger.info(f"query='{query}' -> intent='keywords'")
            return {"type": "keywords", "result": extract_keywords(text)}

        elif "count" in query_lower:
            logger.info(f"query='{query}' -> intent='count'")
            return {"type": "count", "result": word_counter(query)}

        elif "prime" in query_lower:
            logger.info(f"query='{query}' -> intent='prime'")
            number = int(re.findall(r"\d+", query)[0])
            return {"type": "prime", "result": is_prime(number)}

        else:
            logger.info(f"query='{query}' -> intent='general'")
            return {"type": "general", "result": "I can help with calculations, keyword extraction, word counts, and checking if a number is prime. Please rephrase your query."}

    except Exception as e:
        logger.error(f"query='{query}' failed: {e}")
        return {"type": "error", "result": f"Something went wrong: {e}"}
````

## 📦 Expected Output Format

```
{
  "type": "calculation / keywords / general / error",
  "result": ...
}
```

````python
# 🧪 Test Cases

queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is machine learning?",
    "15 * 3",
    "count words in this sentence",
    "is 17 prime"
]

for q in queries:
    print("Query:", q)
    print("Response:", agent(q))
    print("-" * 50)
````

````python
# 🎯 Interactive Mode

while True:
    user_input = input("Enter query (type 'exit' to stop): ")
    if user_input.lower() == "exit":
        break
    print("Response:", agent(user_input))
````
